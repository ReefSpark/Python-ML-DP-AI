## 1.Abstract Method

```python
from abc import ABC,abstractmethod
```

```python
class Bank(ABC):
    @abstractmethod
    def withdraw(self) :
        pass
    def depoist(self) :
        pass
        
```

```python
class IOB(Bank):
    user_name:None
    _amount:None
    def __init__(self,name,amount):
        self.name=name;
        self._amount=amount
    def  withdraw(self,with_amount) :
        self._amount-=with_amount
        
    def  deposit(self,depoist_amount):
        self._amount+=depoist_amount
        
    def showBalance(self):
        print(self._amount) 
        
        
    
```

## 2.Inheritance

```python
class Animal:                                       ## speak function is method overriding
    def __init__(self,name):
        self.name=name
        
    def speak(self):
        print(f" {self.name} animal make a sound")

class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed

    def speak(self) :
        super().speak()
        print("Dervied class")
```
