```python
len(list(mtscar_data['cyl'].value_counts().index))
```

```python
item =[]
explode = []
value = list(mtscar_data['cyl'].value_counts().index)
length = len(list(mtscar_data['cyl'].value_counts().index))
for i in range(0,length) :
    item.append(str(value[i])+" cyl")
    explode.append(0.01)

print(item)
print(explode)
```

```python
pyt.pie(x=mtscar_data['cyl'].value_counts(),labels=item,explode=explode)    ## take value and count and pass a Discrete value
pyt.show()
```

```python
pyt.pie(x=mtscar_data['cyl'].value_counts(),labels=item,explode=explode,autopct='%1.0f%%')       ## autopct show percentage
pyt.show()
```
