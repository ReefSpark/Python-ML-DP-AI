### Positional Arguments

```python
def get_details(name,place,expected_price):
    print(name)
    print(place)
    print(expected_price)
```

```python
get_details(place="chennai",expected_price=25,name="Mathan")
```

### Varaibale length arugment

```python
def get_details(name,place,*expected_price):
    print(name)
    print(place)
    print(expected_price)
```

```python
get_details("chennai","Mathan",59,56.4,45)
```

### Keyword Argument

```python
def get_details1(name,place,**expected_price):
    print(name)
    print(place)
    print(expected_price)
```

```python
get_details1("chennai","Mathan",ask=59,ask2=56.4,ask3=45)
```
