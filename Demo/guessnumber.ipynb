### Try - Catch

```python
import random
secert_number = random.randint(1,100)

for i in range(1,8) :
    try:
        user_input=int(input("Enter the guess number"))
        if(user_input <1 or user_input>100):
             print("Invald number ,Please enter in between 1 to 100")
        elif(user_input>secert_number):
            print("Your number is Higher")
        elif(user_input<secert_number):
            print("Your number is Lesser")
        elif(user_input == secert_number):
               break
        else : 
          print("its worng number")
    except ValueError:
        print("Invalid Character")
if secert_number == user_input:
     print(f"Congrulation user,you find a guess nuber {secert_number} in {i} attempt" )
else:
     print("Try again")
```
