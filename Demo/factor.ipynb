```python
def factor(num) :
    result =1
    if(num<=0):
       print("Number is greater than 0")
    else :
        for i in range(1,num+1):
            result=result*i

    return result
            

  
```

```python
factor(10)
```

```python
def binarySearchValue(data,search:int) -> int:
    list_value=data
    l=0
    r=len(data)-1
    while l<=r :
        mid=(l+r)//2
        if list_value[mid]==search :
            return mid
        elif search>list_value[mid]:
            l=mid+1
        elif search<list_value[mid]:
            r=mid-1
    return -1
```

```python
input_data=[1,2,4,6,8,9,10,34]
print(binarySearchValue(input_data,23))
```
