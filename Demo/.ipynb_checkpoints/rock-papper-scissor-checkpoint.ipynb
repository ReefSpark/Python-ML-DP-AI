```python
import random

def checkCondition(value):
    choseItem = random.choice(["rock", "paper", "scissors"])
    print("Computer chose:", choseItem)

    if (
        (value == "rock" and choseItem == "scissors") or
        (value == "scissors" and choseItem == "paper") or
        (value == "paper" and choseItem == "rock")
    ):
        return 1

    elif value == choseItem:
        return 2

    else:
        return 0


def checkRockPaperScissor(item):
    print(item)
    value = input("Enter your input: ").lower()

    if value in ["rock", "scissors", "paper"]:
        return checkCondition(value)
    else:
        print("Invalid Value")
        return play_again()


def play_again():
    play_again_value = input("Do you want to play again? ").lower()

    if play_again_value == "yes":
        check_output()


def check_output():
    check_value = checkRockPaperScissor("Let's play game")

    if check_value == 1:
        print("You win!")
        play_again()

    elif check_value == 0:
        print("Computer wins!")
        play_again()

    elif check_value == 2:
        print("Match Draw!")
        play_again()


```
