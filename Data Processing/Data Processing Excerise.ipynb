## Find Nan and fill mean value as Solar.R

```python
weather_data[weather_data['Solar.R'].isna()]
```

```python
weather_data.groupby("Weather")['Solar.R'].mean().round()
```

```python
weather_data.head()
```

```python
weather_data['Solar.R']=weather_data.groupby("Weather")['Solar.R'].transform(lambda x:x.fillna(x.mean())).round()
```

```python
weather_data.isna().sum()
```

```python
weather_data[weather_data['Solar.R'].isna()]   ## Output have nan value should be droped
```

```python
weather_data.dropna(inplace=True)
```

```python
weather_data.isna().sum()
```

```python
weather_data.dtypes
```

```python
weather_data['Month'].unique()
```

```python
weather_data['Month'].replace('May','5',inplace=True)
weather_data
```

```python
weather_data['Month']=weather_data['Month'].astype(int)
```

## Stage 2 - Data Transformation
   * if Data is Continous = Standard Scaler,MinMaxScaler, Robust Scaler
   * If data is Discrete = Label Encoder, One hot Encoder

```python
 weather_data_copy_01 = weather_data.copy() 
weather_data_copy_01
```

```python
from sklearn.preprocessing import LabelEncoder       ## weather have 3 value S - Sunny,PS, C -  cold is convert to numberical show an alhabets S - 2 ,PS - 1,C - 0
le_encoder = LabelEncoder()
weather_data_copy_01['Weather']= le_encoder.fit_transform(weather_data_copy_01["Weather"])
weather_data_copy_01
```
