```python
plt.hist(mtcarData['mpg'])   ## speak about the distribution for one continous data
plt.xlabel("Mileage(MPG)")
plt.ylabel("Frequency")
plt.title("MPG Distribution")
plt.show()  ### only show chart
```

```python
value = titanic_data['sex'].value_counts()
print(value)

plt.pie(value,labels=['male','female'] ,autopct="%1.0f%%" ,explode=[0.01,0.01])
plt.show()
```
