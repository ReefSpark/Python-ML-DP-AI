## Create a Module

* login
* view project
* submit project 
* review project 

```python
def  login(user) :
    print(user['name'])
    print( f"login user name as {user['name']}")
    
def view_project(user):
    print(f" {user['name']} is view project")

def  submit(user) :
    if user['role']== 'team memeber' :
        print(f"{user['name']} is submit a project")
    else :
        print("Team Member only access")
        
def  review_project(user) :
    if(user['role'] =='manager') :
        print(f"{user['name']} review a project")
    else :
        print("Manger only access this page")

user_data=[{"name":"Mathan","role":"team member"},{"name":"Rajesh","role":"manager"}]

for user in user_data:
    login(user)
    view_project(user)
    submit(user)
    review_project(user)


        
```

## with  Class

```python
class Employee :

   def __init__(self,name):
     self.name = name
       
   def login(self) :
    print( f" login user name as {self.name}")
       
   def view_project(self):
    print(f" {self.name} is view project")

class TeamMember(Employee):

    def submit(self) :
     print(f"{self.name} is submit a project")
      
class Manager(Employee):

    def review(self) :
     print(f"{self.name} is review a project") 
        
    
team = TeamMember("john")
team.login()
team.view_project()
team.submit()


manager = manager("john")
manager.login()
manager.view_project()
manager.review()

```
