## Note

**Importance of Self & init into the picture

```python
class SecureUser:
    
   def __init__(self,name,atm_pin) :
      self.name=name
      self.atm_pin=atm_pin
       
   def check_pin(self,attempt) :
       return attempt == self.atm_pin
       
   def depoist_cash():
       print("Cash")
       
   def cash_withdraw() :
       pass
```

```python
user=SecureUser("Mathan",1234)
```
