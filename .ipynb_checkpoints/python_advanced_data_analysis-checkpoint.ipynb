```python
import pandas as pd
```

```python
insurance=pd.read_csv("insurance.csv")
insurance
```

### 1.1 Perform Initial Analysis

```python
insurance.isna().sum()
```

```python
insurance.dtypes
```

```python
insurance.describe()
```

### 1.2 Pull up Insurance charge based on region

```python
insurance.groupby(by="region")['charges'].mean().sort_values(ascending=False).round(2) ##round(2) float value get 2 digit
```

### 1.2 Pull up Insurance charge based on region and sex

```python
insurance.groupby(by=["region","sex"])['charges'].max().sort_values(ascending=False).round(2) 
```

## 2.Pivot Table

```python
pd.pivot_table(data=insurance,values="charges",index="region",columns=["sex","children"]).round(2) #presentable comparing 2 categorical feature vs 1 contionus data
```

### 3 Aggregate

#### 3.1 Find out the average insurance charge based ion Region

```python
insurance.groupby(by=['region','sex'])['charges'].agg(func =["mean","min","max","sum"]).round(2)
```

## Cross Tab

#### 4.1 How many Somker and Non Smokers based on region

```python
pd.crosstab(index=insurance['region'],columns=insurance['sex'],margins=True)
```

## 5. Concat

```python
sales_data_2019=pd.read_csv("sales_transaction_2017.csv")
```

```python
sales_data_2018=pd.read_csv("sales_transaction_2018.csv")
```

```python
sales_data_2017=pd.read_csv("sales_transaction_2019.csv")
```

```python
sales_data_editied = pd.concat([sales_data_2017,sales_data_2018,sales_data_2019])
sales_data_editied
```

```python
sales_data_editied.dtypes
```

```python
sales_data_editied=sales_data_editied.drop_duplicates()
sales_data_editied
```

```python
sales_data_editied = sales_data_editied.iloc[:-2]
sales_data_editied
```

```python
sales_data_editied.isna().sum()
```

```python
sales_data_editied['Voucher']=sales_data_editied['Voucher'].str.replace("Sal:","")
```

```python
sales_data_editied
```

```python
sales_data_editied.isna().sum()
```

```python
sales_data_editied['Date'].fillna("")
```

```python
sales_data_editied['Date'].dropna()
```

```python
sales_data_editied.isna().sum()
```

```python
sales_data_editied = sales_data_editied.dropna(subset=['Date','Voucher','Party','Product','Qty','Rate','Gross'])
sales_data_editied
```

```python
sales_data_editied.isna().sum()
```

```python
sales_data_editied.isnull().sum()
sales_data_editied
```
