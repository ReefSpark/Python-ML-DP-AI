```python
def sorted(array):
     q=len(array)
     for i in  range(0,q):
        for j in range(i,q):
            if array[j]<array[i]:
                inter = array[i]
                array[i]=array[j]
                array[j]=inter
     return array

def findMedianSortedArrays(nums1,nums2) -> float:
        m=len(nums1)
        n=len(nums2)
        nums3=[]
        for i in  range(0,m):
            nums3.append(nums1[i])
        for i in  range(0,n):
            nums3.append(nums2[i])
        value = sorted(nums3)
        result = sum(value)/len(value)
        print(result)
```

```python
nums1=[2,4,4]
nums2=[2,2,2,4,4]
findMedianSortedArrays(nums1,nums2)
```
