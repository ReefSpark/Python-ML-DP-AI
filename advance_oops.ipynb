#### Usecase : I want to hide sensitive data so others can't directly access or change it..

```python
class SecureUser:
   name=None
   atm_pin=None
   def check_pin(attempt) :
       return attempt == atm_pin
   def depoist_cash():
       print("Cash")
   def cash_withdraw() :
       pass
```

## Note

**Importance of Self & init into the picture

```python
class SecureUser1:                        ### _  is protected access to derived class 
                                          ### __ is private only access same class
    
   def __init__(self,name,atm_pin) :
      self.name=name
      self._atm_pin=atm_pin
       
   def check_pin(self,attempt) :
       if attempt == self._atm_pin:
         def depoist_cash():
           print("Cash")
         def cash_withdraw() :
           pass
       else :
           return
```
