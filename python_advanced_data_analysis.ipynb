### 1.2 Pull up Insurance charge based on region

```python
insurance.groupby(by="region")['charges'].mean().sort_values(ascending=False).round(2) ##round(2) float value get 2 digit
```

### 1.2 Pull up Insurance charge based on region and sex

```python
insurance.groupby(by=["region","sex"])['charges'].max().sort_values(ascending=False).round(2) 
```

## 2.Pivot Table

```python
pd.pivot_table(data=insurance,values="charges",index="region",columns=["sex","children"]).round(2) #presentable comparing 2 categorical feature vs 1 contionus data
```

### 3 Aggregate

#### 3.1 Find out the average insurance charge based ion Region

```python
insurance.groupby(by=['region','sex'])['charges'].agg(func =["mean","min","max","sum"]).round(2)
```

## Cross Tab

#### 4.1 How many Somker and Non Smokers based on region

```python
pd.crosstab(index=insurance['region'],columns=insurance['sex'],margins=True)
```

## 6 Apply

#### 6.1 Create new column in the same dataframe with the name "Eligiblity Criteria" where if age is >20 say Eligible else not

```python
insurance_data = pd.read_csv("insurance.csv")
insurance_data
```

```python
def get_age(x):
    if x>20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
insurance_data['Eligiblity Criteria']= insurance_data['age'].apply(get_age)
insurance_data
```

## 7.Lambda function
#### Nameless Function

```python
insurance_data["Eligibilty Criteria"] = insurance_data.age.apply(lambda x:"Eligible" if x>20 else "Not Eligible")
insurance_data
```

#### using map,filter,query

```python
insurance_data['age'].map(lambda x:x*1)
```

```python
insurance_data['age'].filter(lambda x:x=="female")
```

```python
insurance_data.filter(items=["age", "sex"])
```

## filter using lambda

```python
filter(lambda x:x=='female',insurance_data["sex"])
```

```python
## return list
```

```python
list(filter(lambda x:x>20,insurance_data["age"]))
```
