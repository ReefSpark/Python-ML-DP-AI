### List
1) Duplicate Value,
2) Muttable,
3) Indexing and Slicing is available

```python
items_to_buy=["Apple","Mango","Pomgranate","Apple",4,5.6,45]
```

```python
items_to_buy[0]
```

```python
items_to_buy[-1]
```

```python
items_to_buy[:3]
```

```python
items_to_buy[2]="Watermelon"
```

```python
items_to_buy.append("Pears")
```

```python
items_to_buy
```

## 4.Dictonary

#### 
1) Key:Value pairs
2) Cannot contain duplicate key but contain duplicate value
3) Mutable
4) Indexing and slicing is not possible

```python
empty = {}
```

```python
type(empty)
```

```python
student_marks={
    "Maths":67,
    "Science":96,
    "Tamil":89
}
    
```

```python
student_marks.get("Maths")
```
