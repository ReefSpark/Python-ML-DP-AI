```python
 class Node:
    def __init__(self, data):
        self.data = data
        self.children = []


class Tree:
    def __init__(self):
        self.root = None

    def add(self, data, parentNode=None):
        newNode = Node(data)

        if self.root is None:
            self.root = newNode
            return

        parent = self.findParentNode(parentNode, self.root)

        if parent is None:
            print("Parent Not found")
            return

        parent.children.append(newNode)

    def findParentNode(self, parentNode, node):
        if node.data == parentNode:
            return node

        for child in node.children:
            nodeFound = self.findParentNode(parentNode, child)
            if nodeFound is not None:
                return nodeFound

        return None

    def display(self, depth=0, node=None):
        if self.root is None:
            print("Tree is empty")
            return

        if node is None:
            node = self.root

        print(" " * depth + str(node.data))

        for child in node.children:
            self.display(depth + 1, child)


# Create the tree
tree = Tree()

# Add nodes
tree.add(1)
tree.add(2, 1)
tree.add(3, 1)
tree.add(4, 2)
tree.add(5, 2)
tree.add(6, 3)

# Display
tree.display()
```
