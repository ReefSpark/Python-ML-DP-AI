```python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []

class Tree:
    def __init__(self):
        self.root = None

    def add(self, data, parentNode=None):
        newNode = Node(data)

        if self.root is None:
            self.root = newNode
            return

        parent = self.findParentNode(parentNode, self.root)

        if parent is None:
            print("Parent Not found")
            return

        parent.children.append(newNode)

    def findParentNode(self, parentNode, node):
        if node.data == parentNode:
            return node
        for child in node.children:
            nodeFound = self.findParentNode(parentNode, child)
            if nodeFound is not None:
                return nodeFound

        return None

    def display(self, depth=0, node=None):
        if node is None:
            node = self.root
        print(" " * depth, node.data)
        for child in node.children:
            self.display(depth + 1, child)



    
            
```

```python
tree = Tree()
tree.add(5)
tree.add(2,5)
tree.add(4,5)
tree.add(6,5)
tree.add(3,2)
tree.add(8,2)
```
