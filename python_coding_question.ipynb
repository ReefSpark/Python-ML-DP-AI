 ### 3.	Write a program that handles division by zero using try-except.

```python
value = int(input("enter you value"))
try:
   if value/0:
     print("Divisible by 0")
except ZeroDivisionError:
     print("Divisible by Error")
finally:
    print("Program End")
```

### 5.	Reverse a list without using built-in functions.

```python
arr = [13,14,78,23,45]
arr2=[]
for i in range((len(arr)-1),-1,-1):
    print(i)
    arr2.append(arr[i])
print(arr2)
```
