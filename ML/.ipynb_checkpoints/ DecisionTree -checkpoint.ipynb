 ## 1.Need Necessary Import

```python
import pandas as pd
```

## 2. Read File

```python
claimants_data = pd.read_csv("claimant.csv")
```

## 3. Data Understaning

```python
claimants_data.shape
```

```python
claimants_data.isna().sum()
```

```python
claimants_data.dropna(inplace=True)
```

## 4.Data Preparation

```python
claimants_data.head()
```

```python
del claimants_data["CASENUM"]
```

```python
x = claimants_data[["CLMSEX","CLMINSUR","SEATBELT","CLMAGE","LOSS"]]
```

```python
y = claimants_data['ATTORNEY']
```

```python
print(x, y)
```

## 5.Model Building

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
X_train ,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.20,shuffle=True,random_state=34)
X_train ,X_test,Y_train,Y_test
```

```python
dtmodel= DecisionTreeClassifier()
```

## 6.Model Training 

```python
dtmodel.fit(X_train,Y_train) 
plt.figure(figsize = (8,10))  
plot_tree(decision_tree=dtmodel,filled=True,fontsize=7)
plt.show()
```

## AdaBoost Classifier,RandomForest Classifier, Voting Classifier

```python
from  sklearn.ensemble import AdaBoostClassifier,RandomForestClassifier,VotingClassifier
```

```python
rf_classifier = RandomForestClassifier(n_estimators=100,max_depth=5)
rf_classifier.fit(X_train,Y_train) 
```

### Stacking Technique

```python
from sklearn.neighbors import KNeighborsClassifier

logisitic_classifier = LogisticRegression()
dt_classifier = DecisionTreeClassifier()
knn_classifier = KNeighborsClassifier


votingClassifier = VotingClassifier(estimators = [("logistic_model",logisitic_classifier),
                                  ("dt_model",dt_classifier),
                                  ("knn_model",knn_classifier)
                                 ])
votingClassifier.fit(x,y)
```

## Grid Search(Cross Valdation)

```python
from sklearn.model_selection  import GridSearchCV
```

```python
grid_search=GridSearchCV(estimator=dtmodel,param_grid={"max_depth":[5,6,7,8,9,10],"criterion":['gini','entropy']})
grid_search
grid_search.fit(x,y)
print(grid_search.best_params_)
print(grid_search.best_score_)
```

## 7.Model Testing 

#### Model Testing Train

```python
y_predict_train = rf_classifier.predict(X_train)
y_predict_train   
```

#### Model Testing Test

```python
y_predict_test = rf_classifier.predict(X_test)
y_predict_test
```

 ## 8. Model Evaluation


```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,roc_auc_score,r2_score,auc,roc_curve
from matplotlib import pyplot as plt
```

#### Training

```python
accuracy_score(Y_train,y_predict_train)   
```

```python
 confusion_matrix(Y_train,y_predict_train) 
```

```python
print(classification_report(Y_train,y_predict_train))
```

### Roc

```python
fpr,tpr,threshold = roc_curve(Y_train,rf_classifier.predict_proba (X_train)[:,1])
auc = roc_auc_score(Y_train,y_predict_train)
print(auc)
plt.plot(fpr,tpr,color='red',label = "Decision Tree (area = %0.2f)" %auc)
plt.plot([0.1],[0.1],'k--')
plt.xlabel("False Postive Rate or [1 - True Negative Rate]")
plt.ylabel('True Positive Rate')
```

#### Testing

```python
accuracy_score(Y_test,y_predict_test) 
```

```python
confusion_matrix(Y_test,y_predict_test) 
```

```python
print(classification_report(Y_test,y_predict_test))
```

### Roc 

```python
fpr,tpr,threshold = roc_curve(Y_test,rf_classifier.predict_proba (X_test)[:,1])
auc = roc_auc_score(Y_test,y_predict_test)
print(auc)
plt.plot(fpr,tpr,color='red',label = "Decision Tree (area = %0.2f)" %auc)
plt.plot([0.1],[0.1],'k--')
plt.xlabel("False Postive Rate or [1 - True Negative Rate]")
plt.ylabel('True Positive Rate')
```
