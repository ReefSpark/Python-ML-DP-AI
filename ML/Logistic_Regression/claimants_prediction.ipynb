## 1.Import Necessacry Libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from matplotlib import pyplot as plt
```

## 2.Import DataSet

```python
claimants_data = pd.read_csv("claimant.csv")
claimants_data
```

## 4. Data Preparation
  * 4.1 Data Cleaning

```python
claimants_data.dropna(inplace=True)
claimants_data
```

```python
claimants_data.isna().sum()
```

#### 4.2 Seperate your inputs and output

```python
x=claimants_data.drop(labels=['CASENUM','ATTORNEY'],axis=1)
x
```

```python
y=claimants_data['ATTORNEY']
y
```

## 5.Model Building 

```python
from sklearn.model_selection import train_test_split
 
X_train, X_test, y_train, y_test=train_test_split(x,y,test_size=0.20,shuffle= True,random_state=34)  ## random_ state I want to get a set of sample data no matter how many time to shuffle a data

print( X_train, X_test, y_train, y_test) 

logistic_model = LogisticRegression()
```

## 6.Model Training

```python
logistic_model.fit(X_train, y_train)
```

## 7.Model Testing

 ### Traning Data

```python
y_predict_train = logistic_model.predict(X_train)
y_predict_train
```

#### Test Data 

```python
y_predict_test=logistic_model.predict(X_test)
y_predict_test
```

```python
claimants_data['Prediction'] = logistic_model.predict(x)
claimants_data
```

```python
y_predict=logistic_model.predict(x)
```

## 8 Model Evaluation

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,auc,roc_auc_score,roc_curve
```

### 8.1 Train Data's Evaluation Metrics

```python
print(confusion_matrix(y_train,y_predict_train))
```

```python
print(classification_report(y_train,y_predict_train))
```

```python
accuracy_score(y_train,y_predict_train)
```

### Roc Curve - Receiver Operating Characteristics
#### AUC Area Under this Curve

```python
fpr,tpr,threshold = roc_curve(y_train,logistic_model.predict_proba (X_train)[:,1])
auc = roc_auc_score(y_train,y_predict_train)
print(auc)
plt.plot(fpr,tpr,color="red",label="logit model (area = %02f)" %auc)
plt.plot([0.1],[0.1],'k--')
plt.xlabel("False Postive Rate or [1 - True Negative Rate]")
plt.ylabel('True Positive Rate')
```

#### Traning Accuary
